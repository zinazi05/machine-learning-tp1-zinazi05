```python
from sklearn.datasets import fetch_20newsgroups
from collections import Counter
import re

# 1-2. Read the text file and print each line
with open("txt_file.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(line.rstrip("\n"))      # removes only the trailing newline
```

```python
# 3. Load 20 Newsgroups (training subset, default categories)
newsgroups = fetch_20newsgroups(subset="train")
```

```python
# 4. Number of documents and labels
print(len(newsgroups.data))           # number of documents
print(len(newsgroups.target))         # number of labels (one per document)
print(newsgroups.target.shape)
```

```python
# 5. First document and its label
print(newsgroups.data[0])
print(newsgroups.target[0])
print(newsgroups.target_names[newsgroups.target[0]])
```

```python
# 6. Class names
print(newsgroups.target_names)
print(len(newsgroups.target_names))
```

```python
# 7. Optional: document lengths and word counts
lengths = [len(doc.split()) for doc in newsgroups.data]
print(sum(lengths) / len(lengths), min(lengths), max(lengths))

words = re.findall(r"[a-z]+", " ".join(newsgroups.data).lower())
print(Counter(words).most_common(10))
```
